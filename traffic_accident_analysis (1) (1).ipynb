Data Cleaning

 * Explanation: Removes duplicate accident records from the dataset.

```python
# Remove duplicate values
ta.drop_duplicates(inplace=True)
uni_rows = ta.shape[0]
print(uni_rows)
```

* Explanation: Resets the dataset index after removing duplicates.

```python
# Reset index
ta.reset_index()
```

* Explanation: Displays missing values present in each column.

```python
# Check missing values
rem_val = ta.isnull().sum().sort_values(ascending=False)
rem_val
```

* Explanation: Calculates the percentage of missing values in the Traffic Control Presence column.

```python
uni_rows = ta.shape[0]
per_traffic = rem_val.iloc[0]/uni_rows*100
print(per_traffic)
```

* Explanation: Calculates the percentage of missing values in the Traffic Control Presence column.

```python
ta['Traffic Control Presence'].fillna(value='Unknown',inplace=True)
ta['Traffic Control Presence'].head()
```

* Missing values in the Traffic Control Presence column were replaced with "Unknown".

```python
ta['Traffic Control Presence'] = ta['Traffic Control Presence'].fillna('Unknown')
ta.isnull().sum()
```

Data Visualization

* Explanation: Displays the number of accidents under each severity category.

```python
ta['Accident Severity'].value_counts().plot(kind='bar')
plt.title('Accident Severity Distribution')
plt.xlabel('Accident Severity')
plt.ylabel('Count')
plt.show()
```

* Explanation: Displays accident frequency under different weather conditions.

```python
ta['Weather Conditions'].value_counts().plot(kind='bar')
plt.title('Weather Conditions Distribution')
plt.xlabel('Weather Conditions')
plt.ylabel('Count')
plt.show()
```

* Explanation: Displays the percentage distribution of accidents by road type.

```python
ta['Road Type'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Road Type Distribution')
plt.ylabel('')
plt.show()
```

* Explanation: Displays the distribution of driver ages involved in accidents.

```python
plt.hist(ta['Driver Age'])
plt.title('Driver Age Distribution')
plt.xlabel('Driver Age')
plt.ylabel('Frequency')
plt.show()
```

* Explanation: Displays total casualties for each accident severity level.

```python
ta.groupby('Accident Severity')['Number of Casualties'].sum().plot(kind='bar')
plt.title('Casualties by Accident Severity')
plt.xlabel('Accident Severity')
plt.ylabel('Total Casualties')
plt.show()
```

```python
ta['Weather Conditions'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Weather Conditions Distribution')
plt.ylabel('')
plt.show()
```

* Explanation: Displays relationships between numerical accident attributes.

```python
numeric_data = ta[['Number of Vehicles Involved',
                   'Number of Casualties',
                   'Number of Fatalities',
                   'Speed Limit (km/h)',
                   'Driver Age']]
plt.figure(figsize=(8,6))
plt.imshow(numeric_data.corr(), cmap='coolwarm')
plt.colorbar()
plt.xticks(range(len(numeric_data.columns)), numeric_data.columns, rotation=90)
plt.yticks(range(len(numeric_data.columns)), numeric_data.columns)
plt.title('Correlation Heatmap')
plt.show()
```

* Explanation: Displays the relationship between driver age and number of casualties.

```python
plt.scatter(
    ta['Driver Age'],
    ta['Number of Casualties']
)
plt.title('Driver Age vs Number of Casualties')
plt.xlabel('Driver Age')
plt.ylabel('Number of Casualties')
plt.show()
```

* Explanation: Displays the number of accidents in each city and helps identify the cities with the highest accident rates.

```python
ta['State Name'].value_counts().plot(kind='bar', figsize=(12,6))
plt.title('Number of Accidents by City')
plt.xlabel('State Name')
plt.ylabel('Number of Accidents')
plt.show()
```

* Explanation: Displays the number of accidents involving alcohol and non-alcohol cases.

```python
ta['Alcohol Involvement'].value_counts().plot(kind='bar')
plt.title('Alcohol Involvement in Accidents')
plt.xlabel('Alcohol Involvement')
plt.ylabel('Number of Accidents')
plt.show()
```

* Explanation:* Among alcohol-related accidents, male drivers were involved more frequently than female drivers.

```python
ta[ta['Alcohol Involvement'] == 'Yes']['Driver Gender'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%'
)
plt.title('Male and Female Drivers with Alcohol Involvement')
plt.ylabel('')
plt.show()
```

** To effectively communicate accident analysis results, Matplotlib is used to create informative and visually appealing visualizations. Heatmaps, pie charts, bar charts, and scatter plots are utilized to present accident severity, casualty distribution, weather conditions, and driver-related patterns. These visualizations help in understanding accident trends and identifying important factors contributing to road accidents.
